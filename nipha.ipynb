```python
#data importing 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df=pd.read_csv("n dataset.csv")
df
```

```python
#cleaning the column names  
df.columns = df.columns.str.strip()
```

```python
#understanding the cases as per the country registered 
ct = pd.crosstab(df["Country"], df["Year"] ,values=df["Cases"], aggfunc="sum")
ct.plot(kind="bar",alpha=0.7,figsize=(5,5))

#given the style to the garph 

plt.xlabel("Country",size=13,color="red",fontweight="bold")
plt.ylabel("Cases Count",size=13,color="red",fontweight="bold")
plt.xticks(size=12,rotation=0)
plt.yticks(size=12)

plt.title("Country vs Cases count",fontweight="bold")

plt.show()
plt.savefig('Cases.png', dpi=300, bbox_inches='tight')
```

```python
#understanding the death as per the country registered 
ct = pd.crosstab(df["Country"], df["Year"] ,values=df["Deaths"], aggfunc="sum")
ct.plot(kind="bar",alpha=0.7,figsize=(5,5))

#given the style to the garph 

plt.xlabel("Country",size=13,color="red",fontweight="bold")
plt.ylabel("Cases Count",size=13,color="red",fontweight="bold")
plt.xticks(size=12,rotation=0)
plt.yticks(size=12)

plt.title("Country vs Death count",fontweight="bold")

plt.show()
plt.savefig('Death.png', dpi=300, bbox_inches='tight')
```

```python
ct = pd.crosstab(df["Year"],df["Mortality Rate (%)"],values=df["Mortality Rate (%)"], aggfunc="sum")
ct.plot(kind="bar",alpha=0.7,figsize=(10,7))

#given the style to the garph 

plt.xlabel("Year",size=13,color="red",fontweight="bold")
plt.ylabel("Mortality Rate (%)",size=13,color="red",fontweight="bold")
plt.xticks(size=12,rotation=0)
plt.yticks(size=12)

plt.title("Country vs Mortality Rate (%)",fontweight="bold")

plt.show()
plt.savefig('molarity rate.png', dpi=300, bbox_inches='tight')
```

```python
#symptoms seen in the patients the percentage of it in horizontal way 
symptoms = [
    "Fever", "Death Symptom", "Lack of Consciousness",
    "Headache", "Vomiting", "Seizures", "Difficulty Breathing"
]

values = [100, 92, 75, 42, 42, 33, 8]
colors = ["red", "green", "blue", "pink", "lime", "violet", "orange"]


plt.figure()
plt.barh(symptoms, values,color=colors,alpha=0.7)

plt.xlabel("Percentage (%)")
plt.ylabel("Symptoms")
plt.title("Symptoms Distribution")

plt.show()
plt.savefig('symptoms_distribution.png', dpi=300, bbox_inches='tight')
```

```python
#data importing 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df=pd.read_csv("nipah_bangladesh_2026.csv")
df
```

```python
#cleaning column names 
df["Date_Confirmed"].str.strip()
```

```python
#removing unwanted columns which are not required
df.drop(["Date_Onset","Date_Admitted","Date_Confirmed","Date_Death"],axis=1,inplace=True)
df
```

```python
#data mapping for easy understanding 
df.replace({"Female":"F","Male":"M"},inplace=True)
df
```

```python
#in this with the help of the bar plot we undersatnd the frequency of the male and the female no. of malesand  get affected by the contact 
ct = pd.crosstab(df["Record_Type"], df["Gender"])
ct.plot(kind="bar",color=("pink","blue"), width=1,alpha=0.7)

plt.xlabel("Record Type",color="red",size=15)
plt.ylabel("Count",color="red",size=15)
plt.title("Case vs Contact by Gender",fontweight="bold")
plt.xticks(rotation=0,size=12)
plt.yticks(size=12)
plt.show()
plt.savefig('symptoms_distribution.png', dpi=300, bbox_inches='tight')
```

```python
#we get to know the no. of people get affected in the particular age group 
ct = pd.crosstab(df["Age_Group"], df["Gender"])
ct.plot(kind="bar",color=("pink","blue"),alpha=0.7)

#given the style to the garph 
plt.style.use("dark_background")
plt.xlabel("Age Group",size=13,color="red",fontweight="bold")
plt.ylabel("Count",size=13,color="red",fontweight="bold")
plt.xticks(size=12,rotation=0)
plt.yticks(size=12)

plt.title("Case vs Contact",fontweight="bold")

plt.show()
```

```python
#we undersatnd the Result of the test as per the age grop 

ct = pd.crosstab(df["Age_Group"], df["Test_Result"])
ct.plot(kind="bar",color=("green","orange","red"),alpha=0.8,figsize=(8,8))

#given the style to the garph 
plt.style.use("dark_background")
plt.xlabel("Age group ",size=12,fontweight="bold",color="blue")
plt.ylabel("Test result count ",size=12,fontweight="bold",color="blue")
plt.xticks(size=12,rotation=0)
plt.yticks(size=12)

plt.title("Test_results as per the Age group  ",fontweight="bold")

plt.show()
plt.savefig('Test_results.png', dpi=300, bbox_inches='tight')
```

```python
#we undersatnd the people get recovered from it
ct = pd.crosstab(df["Test_Result"],df["Final_Status"])
ct.plot(kind="bar",color=("lime","green"),alpha=0.7,figsize=(8,7))

#given the style to the garph 

plt.style.use("dark_background")
plt.xlabel("Test result ",size=13,fontweight="bold",color="red")
plt.ylabel("count",size=13,fontweight="bold",color="red")
plt.xticks(size=12,rotation=0)
plt.yticks(size=12)

plt.title("Test_results as per the Age group  ",fontweight="bold",color="red")

plt.show()
```
